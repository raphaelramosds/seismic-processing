# Convolução — Exercício 9

Se dois sinais discretos e finitos têm $m$ e $n$ amostras, a convolução entre eles terá $m + n - 1$ amostras.

Ver enunciado original em [capitulo1.md](./capitulo1.md).

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 10
m = 15

xn = [0]*n
xm = [0]*m

resultado = np.convolve(xn, xm)

print("xn tem {} elementos".format(len(xn)))
print("xm tem {} elementos".format(len(xm)))
print("Tamanho xn * xm: {} ".format(len(resultado)))
```

## Visualizando o formato da convolução

Os sinais `xn` e `xm` do exemplo acima são compostos só de zeros, então o resultado não é visualmente interessante. Para visualizar melhor o *formato* do resultado da convolução, vamos repetir o experimento com sinais não triviais, mantendo os mesmos tamanhos `m` e `n` para confirmar a fórmula $m+n-1$.

```python
xn_ex = np.zeros(n)
xn_ex[2:5] = [1, 2, 1]  # um pequeno pulso

xm_ex = np.zeros(m)
xm_ex[0:3] = [1, -1, 0.5]  # um filtro simples

resultado_ex = np.convolve(xn_ex, xm_ex)

fig, axs = plt.subplots(3, 1, figsize=(9, 7))

axs[0].stem(xn_ex)
axs[0].set_title(f"xn (m={len(xn_ex)} amostras)")

axs[1].stem(xm_ex)
axs[1].set_title(f"xm (n={len(xm_ex)} amostras)")

axs[2].stem(resultado_ex)
axs[2].set_title(f"xn * xm (m+n-1={len(resultado_ex)} amostras)")

plt.tight_layout()
plt.show()

print("Tamanho esperado (m+n-1):", len(xn_ex) + len(xm_ex) - 1)
print("Tamanho obtido:", len(resultado_ex))
```
